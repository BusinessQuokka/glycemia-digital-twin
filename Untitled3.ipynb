```python
import pandas as pd
import numpy as np
import requests
import time
import logging
from datetime import datetime, timedelta
from dataclasses import dataclass, field, asdict
from typing import List
from pathlib import Path

#  ЛОГИРОВАНИЕ 
logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s | %(levelname)-7s | %(message)s',
    datefmt='%H:%M:%S'
)
logger = logging.getLogger(__name__)

#  КОНФИГУРАЦИЯ 
@dataclass
class Config:
    """Настройки подключения и данных."""
    # Замени на реальный URL 
    backend_url: str = "http://localhost:8000" 
    
    # Данные
    csv_path: str = "ohio_cleaned_full.csv"
    patient_id: int = 559
    
    # Параметры симуляции
    cgm_interval_min: int = 5       # Интервал датчика (5 минут)
    history_batch_size: int = 100   # Размер пакета для истории
    realtime_points: int = 12       # Сколько точек отправить в real-time (12 * 5 = 1 час)
    
    # Тайминги
    batch_delay_sec: float = 0.5    # Пауза между пакетами истории
    realtime_interval_sec: int = 10 # Пауза между точками real-time (для теста 10 сек, для прода 300)
    
    # Надёжность
    request_timeout: int = 10
    max_retries: int = 3

config = Config()
logger.info(f" Конфигурация загружена. Бэкенд: {config.backend_url}")
```

```python
@dataclass
class GlucoseReading:
    """Модель одного показания глюкозы."""
    timestamp: str
    glucose: float
    carbs: float = 0.0
    bolus: float = 0.0

    def to_dict(self):
        return asdict(self)

@dataclass
class HistoryPayload:
    """Пакет исторических данных."""
    patient_id: int
    type: str = "training"
    count: int = 0
    data: List[dict] = field(default_factory=list)

@dataclass
class RealtimePayload:
    """Одно показание реального времени."""
    patient_id: int
    type: str = "testing"
    data: dict = field(default_factory=dict)
```

```python
def refresh_timestamps(df: pd.DataFrame, target_end_time: datetime) -> pd.DataFrame:
    """
    Заменяет старые даты в DataFrame на актуальные, сохраняя интервал.
    Последняя строка станет target_end_time.
    маркер 'Z' для явного указания UTC.
    """
    if df.empty:
        return df
    
    df_sorted = df.sort_values('timestamp').reset_index(drop=True)
    n = len(df_sorted)
    
    # Первая дата = конечная - (кол-во точек * интервал)
    start_time = target_end_time - timedelta(minutes=(n - 1) * config.cgm_interval_min)
    
    # новые даты с суффиксом 'Z' (UTC)
    new_times = [
        (start_time + timedelta(minutes=i * config.cgm_interval_min)).isoformat() + "Z"
        for i in range(n)
    ]
    
    df_result = df_sorted.copy()
    df_result['timestamp'] = new_times
    return df_result

def send_request(endpoint: str, payload: dict) -> bool:
    """Отправляет POST запрос с логикой повторных попыток (retry)."""
    url = f"{config.backend_url}{endpoint}"
    
    for attempt in range(1, config.max_retries + 1):
        try:
            resp = requests.post(url, json=payload, timeout=config.request_timeout)
            resp.raise_for_status()
            logger.info(f" {endpoint} | Status: {resp.status_code} (Attempt {attempt})")
            return True
        except Exception as e:
            logger.warning(f" Ошибка {endpoint} (Attempt {attempt}/{config.max_retries}): {e}")
            if attempt < config.max_retries:
                time.sleep(2)
                
    logger.error(f" Не удалось отправить {endpoint} после {config.max_retries} попыток")
    return False
```

```python
def load_and_prepare_data() -> tuple:
    """Загружает CSV, фильтрует пациента и обновляет даты."""
    
    # Загрузка
    if not Path(config.csv_path).exists():
        raise FileNotFoundError(f"Файл {config.csv_path} не найден!")
        
    df = pd.read_csv(config.csv_path)
    df['timestamp'] = pd.to_datetime(df['timestamp'])
    
    # Фильтрация
    patient_df = df[df['patient_id'] == config.patient_id].copy()
    if patient_df.empty:
        raise ValueError(f"Нет данных для пациента {config.patient_id}")
        
    patient_df = patient_df.sort_values('timestamp').reset_index(drop=True)
    
    # Заполняем пропуски нулями
    for col in ['carbs', 'bolus']:
        if col in patient_df.columns:
            patient_df[col] = patient_df[col].fillna(0.0)
            
    # Разделение на историю и real-time
    realtime_df = patient_df.tail(config.realtime_points).copy()
    history_df = patient_df.iloc[:-config.realtime_points].copy()
    
    # Обновление дат
    now = datetime.utcnow() # Используем UTC для согласованности
    
    # История должна закончиться прямо сейчас
    history_end_time = now
    history_df = refresh_timestamps(history_df, history_end_time)
    
    # Real-time начинается сейчас и идет в будущее
    realtime_start_time = now + timedelta(minutes=config.cgm_interval_min)
    realtime_df = refresh_timestamps(realtime_df, realtime_start_time)
    
    logger.info(f" Данные подготовлены: История={len(history_df)}, RealTime={len(realtime_df)}")
    return history_df, realtime_df

# Запуск подготовки
history_df, realtime_df = load_and_prepare_data()
```

```python
def send_history_data(df: pd.DataFrame):
    """Отправляет историю пакетами на бэкенд."""
    if df.empty:
        logger.info(" История пуста, пропускаем")
        return

    total = len(df)
    logger.info(f" Начало отправки истории ({total} записей)...")
    
    for i in range(0, total, config.history_batch_size):
        batch_df = df.iloc[i:i + config.history_batch_size]
        
        # Преобразуем в список объектов GlucoseReading
        readings = [
            GlucoseReading(
                timestamp=row['timestamp'],
                glucose=float(row['glucose']),
                carbs=float(row.get('carbs', 0.0)),
                bolus=float(row.get('bolus', 0.0))
            ).to_dict()
            for _, row in batch_df.iterrows()
        ]
        
        payload = HistoryPayload(
            patient_id=config.patient_id,
            count=len(readings),
            data=readings
        )
        
        logger.info(f" Отправка пакета {i//config.history_batch_size + 1}...")
        send_request("/api/data/history", asdict(payload))
        
        time.sleep(config.batch_delay_sec)
        
    logger.info(" История успешно отправлена")

# Запуск отправки истории
send_history_data(history_df)
```

```python
def simulate_realtime(df: pd.DataFrame):
    """Отправляет данные по одной точке с интервалом, имитируя живой датчик."""
    if df.empty:
        logger.info(" Нет данных для real-time")
        return

    logger.info(f" Запуск real-time симуляции ({len(df)} точек)...")
    logger.info(" Нажмите кнопку STOP в ноутбуке, чтобы прервать симуляцию")
    
    for idx, (_, row) in enumerate(df.iterrows()):
        # Для real-time всегда берем ТЕКУЩЕЕ время с маркером 'Z'
        current_time_utc = datetime.utcnow().isoformat() + "Z"
        
        reading = GlucoseReading(
            timestamp=current_time_utc,
            glucose=float(row['glucose']),
            carbs=float(row.get('carbs', 0.0)),
            bolus=float(row.get('bolus', 0.0))
        )
        
        payload = RealtimePayload(
            patient_id=config.patient_id,
            data=reading.to_dict()
        )
        
        logger.info(f" Точка {idx+1}/{len(df)}: Glucose={reading.glucose} мг/дл | Time={current_time_utc}")
        send_request("/api/data/realtime", asdict(payload))
        
        # Ждем перед следующей точкой (кроме последней)
        if idx < len(df) - 1:
            time.sleep(config.realtime_interval_sec)
            
    logger.info(" Симуляция real-time завершена")

# Запуск симуляции (можно остановить кнопкой Stop)
simulate_realtime(realtime_df)
```
