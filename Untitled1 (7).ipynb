```python
import pandas as pd
import requests
import time
import json
import os
import logging
from datetime import datetime
from tqdm import tqdm

# Настройка логирования
logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(levelname)s - %(message)s',
    handlers=[
        logging.FileHandler('data_pipeline.log', encoding='utf-8'),
        logging.StreamHandler()
    ]
)

#  НАСТРОЙКИ 
API_URL = "http://localhost:5000/api/data"
BATCH_URL = "http://localhost:5000/api/data/batch"  # Отдельный URL для пакетной отправки
QUEUE_FILE = "unsent_queue.json"
SIMULATOR_DELAY = 1

# Заголовки запросов
HEADERS = {
    "Content-Type": "application/json",
    # "Authorization": "Bearer YOUR_TOKEN_HERE"  # Раскомментируй, если нужна авторизация
}

# Валидация: минимально допустимое значение глюкозы
MIN_GLUCOSE = 20
MAX_GLUCOSE = 600

logging.info(" Библиотеки импортированы")
logging.info(f" API URL: {API_URL}")
logging.info(f" BATCH URL: {BATCH_URL}")
logging.info(f" Файл очереди: {QUEUE_FILE}")
```

```python
def prepare_payload(row):
    """
    Преобразует строку DataFrame в JSON-словарь для отправки.
    Проверяет валидность данных перед отправкой.
    """
    glucose = float(row['glucose']) if pd.notna(row['glucose']) else None
    
    # Валидация глюкозы
    if glucose is not None and (glucose < MIN_GLUCOSE or glucose > MAX_GLUCOSE):
        logging.warning(f" Глюкоза {glucose} вне диапазона {MIN_GLUCOSE}-{MAX_GLUCOSE}")
    
    return {
        "patient_id": str(row['patient_id']),
        "file_type": str(row['file_type']),
        "timestamp": row['timestamp'].isoformat() if pd.notna(row['timestamp']) else None,
        "glucose": glucose,
        "basal_rate": float(row['basal_rate']) if pd.notna(row['basal_rate']) else None,
        "bolus": float(row['bolus']) if pd.notna(row['bolus']) else None,
        "carbs": float(row['carbs']) if pd.notna(row['carbs']) else None,
        "exercise_duration": int(row['exercise_duration']) if pd.notna(row['exercise_duration']) else None,
        "exercise_intensity": int(row['exercise_intensity']) if pd.notna(row['exercise_intensity']) else None
    }

def validate_payload(payload):
    """Проверяет, что payload содержит обязательные поля"""
    required_fields = ['patient_id', 'timestamp', 'glucose']
    for field in required_fields:
        if field not in payload or payload[field] is None:
            return False, f"Отсутствует обязательное поле: {field}"
    return True, "OK"

logging.info(" Функции prepare_payload и validate_payload созданы")
```

```python
def save_to_queue(payload):
    """
    Сохраняет данные в локальный JSON-файл, если сервер недоступен.
    """
    queue = []
    if os.path.exists(QUEUE_FILE):
        try:
            with open(QUEUE_FILE, 'r', encoding='utf-8') as f:
                queue = json.load(f)
        except:
            queue = []
    
    queue.append(payload)
    with open(QUEUE_FILE, 'w', encoding='utf-8') as f:
        json.dump(queue, f, ensure_ascii=False, indent=2)
    logging.info(f" Сохранено в очередь: {payload['timestamp']}")

def send_data(payload, max_retries=3, use_batch=False):
    """
    Отправляет данные на сервер с механизмом повторных попыток.
    use_batch=True использует BATCH_URL вместо API_URL
    """
    # Валидация перед отправкой
    is_valid, message = validate_payload(payload)
    if not is_valid:
        logging.error(f" Неверный payload: {message}")
        return False
    
    url = BATCH_URL if use_batch else API_URL
    
    for attempt in range(max_retries):
        try:
            response = requests.post(
                url, 
                json=payload, 
                headers=HEADERS,
                timeout=30 if use_batch else 5  # Увеличенный таймаут для batch
            )
            
            if response.status_code in [200, 201]:
                logging.info(f" Отправлено: {payload.get('timestamp', 'N/A')}")
                return True
            else:
                logging.warning(f" Ошибка сервера {response.status_code}: {response.text}")
                
        except requests.exceptions.Timeout:
            logging.error(f" Таймаут (попытка {attempt + 1}/{max_retries})")
        except requests.exceptions.ConnectionError as e:
            logging.error(f" Ошибка подключения (попытка {attempt + 1}/{max_retries}): {e}")
        except Exception as e:
            logging.error(f" Неизвестная ошибка: {e}")
        
        time.sleep(2 ** attempt)
    
    save_to_queue(payload)
    return False

logging.info(" Улучшенные функции send_data и save_to_queue созданы")
```

```python
def flush_queue():
    """
    Пытается отправить все накопленные в очереди данные.
    """
    if not os.path.exists(QUEUE_FILE):
        logging.info(" Очередь пуста (файл не существует)")
        return
    
    try:
        with open(QUEUE_FILE, 'r', encoding='utf-8') as f:
            queue = json.load(f)
    except json.JSONDecodeError:
        logging.warning(" Файл очереди поврежден, создаем новый")
        queue = []
    except Exception as e:
        logging.error(f"️ Ошибка чтения очереди: {e}")
        queue = []
    
    if not queue:
        logging.info(" Очередь пуста")
        with open(QUEUE_FILE, 'w', encoding='utf-8') as f:
            json.dump([], f)
        return
    
    logging.info(f" Обработка очереди: {len(queue)} записей...")
    remaining_queue = []
    
    for payload in queue:
        if not send_data(payload):
            remaining_queue.append(payload)
    
    with open(QUEUE_FILE, 'w', encoding='utf-8') as f:
        json.dump(remaining_queue, f, ensure_ascii=False, indent=2)
    
    logging.info(f" Очередь обработана. Осталось: {len(remaining_queue)}")

logging.info(" Функция flush_queue готова")
```

```python
def upload_history():
    """
    Отправляет исторические данные пациента (training выборку).
    """
    logging.info("=" * 60)
    logging.info(" ЗАПУСК ОТПРАВКИ ИСТОРИЧЕСКИХ ДАННЫХ ")
    logging.info("=" * 60)
    
    # Проверка файла
    if not os.path.exists('ohio_cleaned_full.csv'):
        logging.error(" Файл ohio_cleaned_full.csv не найден!")
        raise FileNotFoundError("Файл ohio_cleaned_full.csv не найден!")
    
    flush_queue()
    
    df = pd.read_csv('ohio_cleaned_full.csv')
    df['timestamp'] = pd.to_datetime(df['timestamp'])
    df_training = df[df['file_type'] == 'training']
    
    logging.info(f" Найдено {len(df_training)} training записей")
    
    batch = []
    sent_count = 0
    failed_count = 0
    
    # Используем tqdm для прогресс-бара
    for idx, row in tqdm(df_training.iterrows(), total=len(df_training), desc="Обработка"):
        payload = prepare_payload(row)
        batch.append(payload)
        
        if len(batch) >= 100:
            try:
                response = requests.post(
                    BATCH_URL,
                    json=batch,
                    headers=HEADERS,
                    timeout=30
                )
                if response.status_code in [200, 201]:
                    sent_count += len(batch)
                    logging.info(f" Пакет отправлен: {len(batch)} записей (всего: {sent_count})")
                else:
                    logging.warning(f" Ошибка пакета: {response.status_code}")
                    for p in batch:
                        save_to_queue(p)
                    failed_count += len(batch)
            except Exception as e:
                logging.error(f" Ошибка отправки пакета: {e}")
                for p in batch:
                    save_to_queue(p)
                failed_count += len(batch)
            
            batch = []
    
    # Отправка остатка
    if batch:
        try:
            response = requests.post(BATCH_URL, json=batch, headers=HEADERS, timeout=30)
            if response.status_code in [200, 201]:
                sent_count += len(batch)
                logging.info(f" Последний пакет: {len(batch)} записей")
            else:
                for p in batch:
                    save_to_queue(p)
                failed_count += len(batch)
        except Exception as e:
            logging.error(f" Ошибка: {e}")
            for p in batch:
                save_to_queue(p)
            failed_count += len(batch)
    
    logging.info("=" * 60)
    logging.info(" ИСТОРИЧЕСКИЕ ДАННЫЕ ОТПРАВЛЕНЫ")
    logging.info(f" Успешно: {sent_count}")
    logging.info(f" Неудачно: {failed_count}")
    logging.info("=" * 60)

logging.info(" Функция upload_history с прогресс-баром готова")
```

```python
def run_simulator():
    """
    Симулятор данных - имитирует поток данных от реального пациента.
    """
    logging.info("=" * 60)
    logging.info(" ЗАПУСК СИМУЛЯТОРА ДАННЫХ (TESTING)")
    logging.info("=" * 60)
    
    if not os.path.exists('ohio_cleaned_full.csv'):
        logging.error(" Файл ohio_cleaned_full.csv не найден!")
        raise FileNotFoundError("Файл ohio_cleaned_full.csv не найден!")
    
    df = pd.read_csv('ohio_cleaned_full.csv')
    df['timestamp'] = pd.to_datetime(df['timestamp'])
    df_testing = df[df['file_type'] == 'testing'].sort_values('timestamp')
    
    logging.info(f" Симуляция {len(df_testing)} записей")
    logging.info(f" Задержка: {SIMULATOR_DELAY} сек")
    logging.info(f" Примерное время: ~{len(df_testing) * SIMULATOR_DELAY / 60:.1f} мин\n")
    
    sent_count = 0
    failed_count = 0
    
    for idx, row in tqdm(df_testing.iterrows(), total=len(df_testing), desc="Симуляция"):
        payload = prepare_payload(row)
        
        if send_data(payload):
            sent_count += 1
        else:
            failed_count += 1
        
        if (idx + 1) % 10 == 0:
            logging.info(f"  [{payload['timestamp']}] Глюкоза: {payload['glucose']} | "
                        f"{sent_count} | {failed_count}")
        
        time.sleep(SIMULATOR_DELAY)
    
    logging.info("=" * 60)
    logging.info(" СИМУЛЯЦИЯ ЗАВЕРШЕНА")
    logging.info(f" Успешно: {sent_count}")
    logging.info(f" Ошибок: {failed_count}")
    logging.info("=" * 60)

logging.info(" Функция run_simulator с tqdm готова")
```

```python
def load_and_prepare_data(mode='realtime'):
    """
    mode='history': Фильтрует training set для 2-х существующих пациентов,
                    сдвигает время так, чтобы ПОСЛЕДНЯЯ запись = момент запуска, и сохраняет в файл.
    mode='realtime': Фильтрует testing set для 4-х пациентов, сортирует по времени,
                      сдвигает так, чтобы ПЕРВАЯ запись = текущий момент, и сохраняет в файл.
    """
    # 1.КОНФИГУРАЦИЯ 4-х ПАЦИЕНТОВ 
    EXISTING_USERS = [559, 563]  # 2 пациента: получат ИСТОРИЮ + REAL-TIME тесты
    NEW_USERS = [570, 575]       # 2 пациента: получат ТОЛЬКО REAL-TIME тесты (как новые)
    ALL_USERS = EXISTING_USERS + NEW_USERS
    
    # Загрузка исходного CSV
    df = pd.read_csv('ohio_cleaned_full.csv')
    df['timestamp'] = pd.to_datetime(df['timestamp'])
    
    if mode == 'history':
        # Только training и только существующие пользователи
        result_df = df[(df['file_type'] == 'training') & (df['patient_id'].isin(EXISTING_USERS))].copy()
        result_df = result_df.sort_values('timestamp').reset_index(drop=True)
        
        # СДВИГ: ПОСЛЕДНЯЯ запись истории = момент запуска
        max_date = result_df['timestamp'].max()
        if max_date.tz is None:
            max_date = max_date.tz_localize('UTC')
        
        now = pd.Timestamp.now(tz='UTC')
        time_shift = now - max_date
        result_df['timestamp'] = result_df['timestamp'] + time_shift
        if result_df['timestamp'].dt.tz is None:
            result_df['timestamp'] = result_df['timestamp'].dt.tz_localize('UTC')
        
        # СОХРАНЕНИЕ В ФАЙЛ 
        result_df.to_csv('ohio_history_filtered.csv', index=False)
        logging.info(f" Режим ИСТОРИИ: {len(result_df)} записей для {EXISTING_USERS}")
        logging.info(" Последняя запись истории установлена на текущий момент (UTC)")
        logging.info(" Сохранено в файл: ohio_history_filtered.csv")
        
    else:  # mode == 'realtime'
        # Только testing и все 4 пользователя, отсортировано по времени
        result_df = df[(df['file_type'] == 'testing') & (df['patient_id'].isin(ALL_USERS))].copy()
        result_df = result_df.sort_values('timestamp').reset_index(drop=True)
        
        # СДВИГ: ПЕРВАЯ запись = момент запуска
        min_date = result_df['timestamp'].min()
        if min_date.tz is None:
            min_date = min_date.tz_localize('UTC')
        
        now = pd.Timestamp.now(tz='UTC')
        time_shift = now - min_date
        result_df['timestamp'] = result_df['timestamp'] + time_shift
        if result_df['timestamp'].dt.tz is None:
            result_df['timestamp'] = result_df['timestamp'].dt.tz_localize('UTC')
        
        # СОХРАНЕНИЕ В ФАЙЛ
        result_df.to_csv('ohio_realtime_updated.csv', index=False)
        logging.info(f" Режим REAL-TIME: {len(result_df)} записей для {ALL_USERS}")
        logging.info(" Первая запись установлена на текущий момент (UTC)")
        logging.info(" Сохранено в файл: ohio_realtime_updated.csv")
    
    return result_df
```

```python
# ЭТАП 1: ЗАГРУЗКА И ОТПРАВКА ИСТОРИИ 
def check_server_connection():
    """
    Быстрая проверка, что сервер вообще доступен, прежде чем гонять по нему тысячи записей.
    """
    try:
        response = requests.get(API_URL.replace("/readings", ""), timeout=5)
        logging.info(f"Сервер отвечает, статус: {response.status_code}")
        return True
    except requests.exceptions.ConnectionError as e:
        logging.error(f"Сервер недоступен по адресу {API_URL}: {e}")
        return False
    except requests.exceptions.Timeout:
        logging.error(f"Сервер не отвечает (таймаут) по адресу {API_URL}")
        return False
    except Exception as e:
        logging.warning(f"Не удалось проверить сервер (продолжаем): {e}")
        return True


def send_batch(batch, sent_count, failed_count):
    """
    Отправляет один пакет и возвращает обновленные счетчики.
    Теперь логирует реальную причину ошибки вместо молчаливой отправки в очередь.
    """
    try:
        response = requests.post(BATCH_URL, json=batch, headers=HEADERS, timeout=30)
        if response.status_code in [200, 201]:
            sent_count += len(batch)
            logging.info(f"Пакет отправлен: {len(batch)} записей (всего: {sent_count})")
        else:
            logging.error(
                f"Пакет отклонен сервером. Статус: {response.status_code}. "
                f"Тело ответа: {response.text[:500]}"
            )
            for p in batch:
                save_to_queue(p)
            failed_count += len(batch)
    except requests.exceptions.ConnectionError as e:
        logging.error(f"Не удалось подключиться к серверу: {e}")
        for p in batch:
            save_to_queue(p)
        failed_count += len(batch)
    except requests.exceptions.Timeout:
        logging.error("Таймаут при отправке пакета")
        for p in batch:
            save_to_queue(p)
        failed_count += len(batch)
    except Exception as e:
        logging.error(f"Ошибка пакета: {e}")
        for p in batch:
            save_to_queue(p)
        failed_count += len(batch)

    return sent_count, failed_count


def run_history_upload():
    logging.info("=" * 60)
    logging.info("ЭТАП 1: ЗАГРУЗКА ИСТОРИИ (Запускать ОДИН РАЗ)")
    logging.info("=" * 60)

    if not check_server_connection():
        logging.error("Сервер недоступен, отправка истории прервана. Все записи уйдут в очередь.")

    history_df = load_and_prepare_data(mode='history')

    batch = []
    sent_count = 0
    failed_count = 0
    for idx, row in tqdm(history_df.iterrows(), total=len(history_df), desc="История"):
        payload = prepare_payload(row)
        batch.append(payload)

        if len(batch) >= 100:
            sent_count, failed_count = send_batch(batch, sent_count, failed_count)
            batch = []

    if batch:
        sent_count, failed_count = send_batch(batch, sent_count, failed_count)

    logging.info(f"История обработана. Успешно: {sent_count}. Неудачно: {failed_count}")
```

```python
# ЭТАП 2: REAL-TIME СИМУЛЯЦИЯ 
def run_realtime_simulation():
    """
    ЭТАП 2: REAL-TIME СИМУЛЯЦИЯ.
    Отправляет каждую запись с задержкой, равной реальному интервалу
    между её timestamp и timestamp предыдущей записи, так что первая
    запись уходит сразу (= момент запуска), а дальше поток идет
    в реальном темпе исходных данных.
    """
    logging.info("=" * 60)
    logging.info(" ЭТАП 2: REAL-TIME СИМУЛЯЦИЯ (Запускать НА ВСТРЕЧЕ) ")
    logging.info("=" * 60)
    
    # 1. Получаем отсортированные и сдвинутые данные (первая запись = сейчас)
    realtime_df = load_and_prepare_data(mode='realtime')
    
    # 2. Пошаговая отправка с задержкой, соответствующей реальным интервалам
    sent_count = 0
    failed_count = 0
    prev_timestamp = None
    
    for idx, row in tqdm(realtime_df.iterrows(), total=len(realtime_df), desc="Симуляция"):
        current_timestamp = row['timestamp']
        
        if prev_timestamp is not None:
            gap_seconds = (current_timestamp - prev_timestamp).total_seconds()
            if gap_seconds > 0:
                time.sleep(gap_seconds)
        
        payload = prepare_payload(row)
        if send_data(payload):
            sent_count += 1
        else:
            failed_count += 1
        
        prev_timestamp = current_timestamp
    
    logging.info(f"Симуляция завершена. Успешно: {sent_count}, Ошибок: {failed_count}")


run_history_upload()
# run_realtime_simulation()
```
