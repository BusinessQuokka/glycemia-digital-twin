```python
import json
import numpy as np
from datetime import datetime, timedelta
import pandas as pd
import logging

# Настройка логирования
logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(levelname)s - %(message)s',
    handlers=[
        logging.StreamHandler(),
        logging.FileHandler('medical_logic.log', encoding='utf-8')
    ]
)

# КЛИНИЧЕСКИЕ ГРАНИЦЫ (по умолчанию) 
DEFAULT_GLUCOSE_CRITICAL_LOW = 70      # мг/дл
DEFAULT_GLUCOSE_LOW_RISK = 90          # мг/дл
DEFAULT_GLUCOSE_TARGET_MIN = 90        # мг/дл
DEFAULT_GLUCOSE_TARGET_MAX = 160       # мг/дл
DEFAULT_GLUCOSE_HIGH_RISK = 180        # мг/дл
DEFAULT_GLUCOSE_CRITICAL_HIGH = 250    # мг/дл

# Пороги тренда (мг/дл в минуту)
TREND_FALLING_FAST = -2.0
TREND_FALLING_SLOW = -0.5
TREND_RISING_SLOW = 0.5
TREND_RISING_FAST = 2.0

# Физиологические границы для валидации
GLUCOSE_MIN_PHYSIOLOGICAL = 20   # мг/дл
GLUCOSE_MAX_PHYSIOLOGICAL = 600  # мг/дл

# Окна анализа
TREND_WINDOW_MIN = 30  # последние 30 минут для тренда
DATA_WINDOW_MIN = 60   # последний час для контекста

logging.info(" Библиотеки импортированы, клинические границы заданы")
```

```python
TEMPLATES = {
    'T1': {
        'text': " КРИТИЧЕСКИ НИЗКИЙ УРОВЕНЬ! Глюкоза {GLUCOSE} мг/дл. Немедленно примите 15-20 г быстрых углеводов (сок, таблетки глюкозы).",
        'urgency': 'high'
    },
    'T2': {
        'text': "️ Риск гипогликемии. Глюкоза {GLUCOSE} мг/дл и быстро падает ({TREND}). Съешьте 10-15 г углеводов.",
        'urgency': 'high'
    },
    'T3': {
        'text': " Ночное снижение. Глюкоза {GLUCOSE} мг/дл. Тренд: {TREND}. Модель прогнозирует {PREDICTION} мг/дл через {HORIZON}. Рекомендуется съесть медленные углеводы перед сном.",
        'urgency': 'high'
    },
    'T4': {
        'text': " КРИТИЧЕСКИ ВЫСОКИЙ УРОВЕНЬ! Глюкоза {GLUCOSE} мг/дл. Проверьте кетоны и введите корректирующий болюс.",
        'urgency': 'high'
    },
    'T5': {
        'text': " Риск гипергликемии. Глюкоза {GLUCOSE} мг/дл и быстро растет ({TREND}). Модель прогнозирует {PREDICTION} мг/дл через {HORIZON}. Проверьте, не забыли ли вы инсулин на еду.",
        'urgency': 'medium'
    },
    'T6': {
        'text': " Постпрандиальный рост. Глюкоза {GLUCOSE} мг/дл после приема пищи. Если через 2 часа не снизится до {TARGET} мг/дл, рассмотрите легкую прогулку.",
        'urgency': 'low'
    },
    'T7': {
        'text': " Глюкоза {GLUCOSE} мг/дл (норма), но наблюдается рост ({TREND}). Избегайте дополнительных перекусов.",
        'urgency': 'low'
    },
    'T8': {
        'text': " Глюкоза {GLUCOSE} мг/дл (норма), но наблюдается снижение ({TREND}). Рекомендуется небольшой перекус.",
        'urgency': 'low'
    },
    'T9': {
        'text': " Отличное состояние. Глюкоза {GLUCOSE} мг/дл. Тренд стабильный. Продолжайте соблюдать план.",
        'urgency': 'low'
    },
    'T10': {
        'text': " Анализ: средняя глюкоза {GLUCOSE} мг/дл за 30 мин. Тренд: {TREND}. Модель прогнозирует {PREDICTION} мг/дл через {HORIZON}.",
        'urgency': 'low'
    }
}

logging.info(f" База шаблонов загружена: {len(TEMPLATES)} сценариев")
```

```python
def validate_glucose(value):
    """Проверяет, что значение глюкозы в физиологических границах"""
    if value is None or np.isnan(value):
        return False
    return GLUCOSE_MIN_PHYSIOLOGICAL <= value <= GLUCOSE_MAX_PHYSIOLOGICAL

def calculate_trend(glucose_array):
    """
    Вычисляет тренд в мг/дл в минуту.
    Обработка NaN: пропускает невалидные значения.
    """
    # Фильтрация NaN
    valid_values = [v for v in glucose_array if validate_glucose(v)]
    
    if len(valid_values) < 2:
        logging.warning("️ Недостаточно валидных точек для расчёта тренда")
        return 0.0
    
    # Используем первое и последнее валидное значение
    # Предполагаем, что точки идут с интервалом 5 минут
    time_minutes = (len(valid_values) - 1) * 5
    if time_minutes == 0:
        return 0.0
    
    roc = (valid_values[-1] - valid_values[0]) / time_minutes
    return round(roc, 2)

def get_trend_text(roc):
    """Переводит числовой тренд в текст"""
    if roc < TREND_FALLING_FAST:
        return "резко падает"
    elif roc < TREND_FALLING_SLOW:
        return "снижается"
    elif roc <= TREND_RISING_SLOW:
        return "стабилен"
    elif roc <= TREND_RISING_FAST:
        return "растет"
    else:
        return "резко растет"

def is_night_time(timestamp_str):
    """Проверяет ночное время (00:00 - 06:00)"""
    try:
        hour = datetime.fromisoformat(timestamp_str).hour
        return 0 <= hour < 6
    except:
        return False

def aggregate_context(readings, timestamp_str):
    """
    Извлекает контекст из массива чтений.
    динамически определяет 30-минутное окно по timestamp, а не по индексу.
    """
    try:
        current_time = datetime.fromisoformat(timestamp_str)
    except:
        logging.error(" Невалидный timestamp")
        return [], 0, 0
    
    # Определяем границу 30 минут назад
    trend_cutoff = current_time - timedelta(minutes=TREND_WINDOW_MIN)
    
    glucose_30min = []
    carbs_last_hour = 0
    bolus_last_hour = 0
    
    # Обрезаем readings до последнего часа (если их больше 12)
    relevant_readings = readings[-12:] if len(readings) > 12 else readings
    
    for i, reading in enumerate(relevant_readings):
        # Вычисляем время для этой точки (предполагаем 5-минутный интервал)
        # Последняя точка = current_time, предпоследняя = current_time - 5min, и т.д.
        point_time = current_time - timedelta(minutes=(len(relevant_readings) - 1 - i) * 5)
        
        glucose = reading.get('glucose')
        if validate_glucose(glucose):
            # Если точка попадает в 30-минутное окно
            if point_time >= trend_cutoff:
                glucose_30min.append(glucose)
        
        # Суммируем углеводы и болюс за весь час
        carbs_last_hour += reading.get('carbs', 0) or 0
        bolus_last_hour += reading.get('bolus', 0) or 0
    
    if len(glucose_30min) < 2:
        logging.warning(f" Мало точек в 30-минутном окне: {len(glucose_30min)}")
    
    return glucose_30min, carbs_last_hour, bolus_last_hour

def analyze_predictions(predictions):
    """
    Анализирует предсказания от обеих моделей.
    проверяет направление тренда, а не только абсолютную разницу.
    """
    pred_30min = None
    pred_60min = None
    confidence_30min = 0
    models_disagree = False
    
    # Собираем предсказания по горизонтам
    pred_30_values = []
    pred_60_values = []
    
    for pred in predictions:
        horizon = pred.get('horizon', '')
        glucose = pred.get('glucose')
        confidence = pred.get('confidence', 0)
        
        # Валидация предсказания
        if not validate_glucose(glucose):
            logging.warning(f" Невалидное предсказание: {glucose}")
            continue
        
        if '30' in horizon:
            pred_30_values.append({'glucose': glucose, 'confidence': confidence})
            if confidence > confidence_30min:
                pred_30min = glucose
                confidence_30min = confidence
        
        if '60' in horizon:
            pred_60_values.append(glucose)
            if pred_60min is None:
                pred_60min = glucose
    
    # Проверка расхождения моделей
    if len(pred_30_values) >= 2:
        glucose_values = [p['glucose'] for p in pred_30_values]
        max_diff = max(glucose_values) - min(glucose_values)
        
        # Проверяем направление: все модели предсказывают рост или падение?
        current_glucose = pred_30_values[0]['glucose']  # примерно
        directions = []
        for p in pred_30_values:
            if p['glucose'] > current_glucose + 10:
                directions.append('rising')
            elif p['glucose'] < current_glucose - 10:
                directions.append('falling')
            else:
                directions.append('stable')
        
        # Если модели расходятся по направлению ИЛИ разница > 20 мг/дл
        if len(set(directions)) > 1 or max_diff > 20:
            models_disagree = True
            logging.info(f"️ Расхождение моделей: разница {max_diff:.1f} мг/дл, направления {directions}")
    
    return pred_30min, pred_60min, models_disagree

logging.info(" Вспомогательные функции созданы ")
```

```python
def generate_medical_recommendation(glucose_30min, timestamp_str, 
                                     carbs_last_hour=0, bolus_last_hour=0,
                                     pred_30min=None, pred_60min=None,
                                     models_disagree=False,
                                     patient_profile=None):
    """
    Детерминированный алгоритм выбора рекомендации.
    

    - Персонализация границ через patient_profile
    - Учёт расхождения моделей в логике выбора
    - Более детальное логирование
    """
    # Персонализация границ 
    if patient_profile:
        critical_low = patient_profile.get('critical_low', DEFAULT_GLUCOSE_CRITICAL_LOW)
        low_risk = patient_profile.get('low_risk', DEFAULT_GLUCOSE_LOW_RISK)
        target_min = patient_profile.get('target_min', DEFAULT_GLUCOSE_TARGET_MIN)
        target_max = patient_profile.get('target_max', DEFAULT_GLUCOSE_TARGET_MAX)
        high_risk = patient_profile.get('high_risk', DEFAULT_GLUCOSE_HIGH_RISK)
        critical_high = patient_profile.get('critical_high', DEFAULT_GLUCOSE_CRITICAL_HIGH)
    else:
        critical_low = DEFAULT_GLUCOSE_CRITICAL_LOW
        low_risk = DEFAULT_GLUCOSE_LOW_RISK
        target_min = DEFAULT_GLUCOSE_TARGET_MIN
        target_max = DEFAULT_GLUCOSE_TARGET_MAX
        high_risk = DEFAULT_GLUCOSE_HIGH_RISK
        critical_high = DEFAULT_GLUCOSE_CRITICAL_HIGH
    
    current_glucose = round(glucose_30min[-1]) if glucose_30min else 0
    trend_roc = calculate_trend(glucose_30min)
    trend_text = get_trend_text(trend_roc)
    
    # Определяем прогноз и горизонт
    prediction_value = pred_60min if pred_60min else pred_30min
    horizon_text = "60 мин" if pred_60min else "30 мин"
    
    # Базовый urgency
    base_urgency = 'medium' if models_disagree else 'low'
    
    #  ЛОГИКА ВЫБОРА ШАБЛОНА 
    
    # 1. Критическая гипогликемия
    if current_glucose < critical_low:
        template_id = 'T1'
        urgency = 'high'
        logging.info(f" Критическая гипо: {current_glucose} мг/дл")
        
    # 2. Ожидаемый переход за нижнюю границу
    elif critical_low <= current_glucose < low_risk and trend_roc < -1.0:
        if is_night_time(timestamp_str):
            template_id = 'T3'
        else:
            template_id = 'T2'
        urgency = 'high'
        logging.info(f"️ Риск гипо: {current_glucose} мг/дл, тренд {trend_roc}")
        
    # 3. Критическая гипергликемия
    elif current_glucose > critical_high:
        template_id = 'T4'
        urgency = 'high'
        logging.info(f" Критическая гипер: {current_glucose} мг/дл")
        
    # 4. Ожидаемый переход за верхнюю границу
    elif high_risk < current_glucose <= critical_high and trend_roc > 2.0:
        if carbs_last_hour > 0:
            template_id = 'T6'
        else:
            template_id = 'T5'
        urgency = 'medium'
        logging.info(f" Риск гипер: {current_glucose} мг/дл, тренд {trend_roc}")
        
    # 5. Целевой диапазон
    elif target_min <= current_glucose <= target_max:
        if trend_roc > 1.0:
            template_id = 'T7'
        elif trend_roc < -1.0:
            template_id = 'T8'
        else:
            template_id = 'T9'
        urgency = base_urgency
        
    # 6. Фолбэк
    else:
        template_id = 'T10'
        urgency = base_urgency
    
    # ПОДСТАНОВКА ПАРАМЕТРОВ 
    template = TEMPLATES[template_id]
    recommendation_text = template['text'].format(
        GLUCOSE=current_glucose,
        TREND=trend_text,
        PREDICTION=round(prediction_value) if prediction_value else "N/A",
        HORIZON=horizon_text,
        TARGET=target_max
    )
    
    # Если urgency не был явно задан — берем из шаблона
    if urgency == 'low' and template['urgency'] != 'low':
        urgency = template['urgency']
    
    return template_id, recommendation_text, urgency

logging.info(" Главная функция детерминированного алгоритма создана ")
```

```python
def process_recommendation_request(request_json, patient_profile=None):
    """
    Принимает JSON от бэкенда в согласованном формате.
    
   
    - Полная валидация входных данных
    - Обработка ошибок с graceful degradation
    - Детальное логирование
    """
    try:
        # Валидация структуры
        if 'timestamp' not in request_json:
            raise ValueError("Отсутствует поле 'timestamp'")
        if 'readings' not in request_json:
            raise ValueError("Отсутствует поле 'readings'")
        if not isinstance(request_json['readings'], list):
            raise ValueError("Поле 'readings' должно быть массивом")
        if len(request_json['readings']) == 0:
            raise ValueError("Массив 'readings' пуст")
        
        # Извлечение данных
        timestamp = request_json['timestamp']
        readings = request_json['readings']
        predictions = request_json.get('predictions', [])
        
        logging.info(f" Получен запрос: {len(readings)} readings, {len(predictions)} predictions")
        
        # Агрегация контекста
        glucose_30min, carbs_last_hour, bolus_last_hour = aggregate_context(readings, timestamp)
        
        if len(glucose_30min) < 2:
            logging.warning(" Недостаточно данных для тренда, используем fallback")
            glucose_30min = [readings[-1].get('glucose', 100)] * 6
        
        # Анализ предсказаний
        pred_30min, pred_60min, models_disagree = analyze_predictions(predictions)
        
        # Вызов детерминированного алгоритма
        template_id, recommendation_text, urgency = generate_medical_recommendation(
            glucose_30min=glucose_30min,
            timestamp_str=timestamp,
            carbs_last_hour=carbs_last_hour,
            bolus_last_hour=bolus_last_hour,
            pred_30min=pred_30min,
            pred_60min=pred_60min,
            models_disagree=models_disagree,
            patient_profile=patient_profile
        )
        
        # Формирование ответа
        response = {
            'template_id': template_id,
            'urgency': urgency,
            'recommendation_text': recommendation_text,
            'metadata': {
                'current_glucose': round(glucose_30min[-1]) if glucose_30min else None,
                'trend_roc': calculate_trend(glucose_30min),
                'models_disagree': models_disagree,
                'prediction_30min': pred_30min,
                'prediction_60min': pred_60min
            }
        }
        
        logging.info(f" Рекомендация сгенерирована: {template_id} (urgency: {urgency})")
        return response
        
    except Exception as e:
        logging.error(f" Ошибка обработки запроса: {e}", exc_info=True)
        return {
            'template_id': 'ERROR',
            'urgency': 'high',
            'recommendation_text': f"Ошибка обработки данных: {str(e)}",
            'metadata': {'error': str(e)}
        }

logging.info(" Функция-обёртка для бэкенда готова ")
```

```python
# Загрузка реальных данных
df = pd.read_csv('ohio_cleaned_full.csv')
df['timestamp'] = pd.to_datetime(df['timestamp'])

# Берем данные пациента 559
patient_id = 559
patient_data = df[df['patient_id'] == patient_id].sort_values('timestamp')

# Берем последние 12 точек (60 минут)
last_60min = patient_data.tail(12)

# Формируем readings
readings = []
for _, row in last_60min.iterrows():
    readings.append({
        'glucose': float(row['glucose']) if pd.notna(row['glucose']) else None,
        'carbs': float(row['carbs']) if pd.notna(row['carbs']) else 0,
        'bolus': float(row['bolus']) if pd.notna(row['bolus']) else 0
    })

# Формируем тестовый запрос
test_request = {
    'timestamp': last_60min['timestamp'].iloc[-1].isoformat(),
    'readings': readings,
    'predictions': [
        {'model': 'nn', 'horizon': '30min', 'glucose': readings[-1]['glucose'] + 10, 'confidence': 0.85},
        {'model': 'ode', 'horizon': '30min', 'glucose': readings[-1]['glucose'] + 8, 'confidence': 0.90}
    ]
}

print(" Тестовые данные:")
print(f"Пациент: {patient_id}")
print(f"Время: {test_request['timestamp']}")
print(f"Глюкоза за 30 мин: {[r['glucose'] for r in readings[-6:]]}")
print(f"Углеводы за час: {sum(r['carbs'] for r in readings)}")
print(f"Болюс за час: {sum(r['bolus'] for r in readings)}")
print("\n" + "="*60)

# Вызов алгоритма
result = process_recommendation_request(test_request)

print("\n РЕЗУЛЬТАТ:")
print(f"Шаблон: {result['template_id']}")
print(f"Срочность: {result['urgency']}")
print(f"Рекомендация: {result['recommendation_text']}")
print(f"\nМетаданные: {json.dumps(result['metadata'], indent=2, ensure_ascii=False)}")
```

```python
#  ИНТЕГРАЦИЯ С БЭКЕНДОМ 

import requests
import time
from dataclasses import dataclass, asdict

# Добавляем функцию отправки 
def send_recommendation_to_backend(payload: dict, backend_url: str = "http://localhost:8000") -> bool:
    """Отправляет готовую рекомендацию на бэкенд."""
    endpoint = "/api/recommendations"  # ЗАМЕНИ 
    url = f"{backend_url}{endpoint}"
    
    for attempt in range(1, 4):
        try:
            resp = requests.post(url, json=payload, timeout=10)
            resp.raise_for_status()
            print(f" Рекомендация отправлена! Status: {resp.status_code}")
            return True
        except Exception as e:
            print(f" Ошибка отправки (попытка {attempt}/3): {e}")
            if attempt < 3:
                time.sleep(2)
    return False

# Создаём тестовые данные 
test_input = {
    "patient_id": 559,
    "timestamp": "2026-07-19T12:00:00Z",
    "readings": [
        {"glucose": 110.0, "carbs": 0.0, "bolus": 0.0},
        {"glucose": 115.0, "carbs": 0.0, "bolus": 0.0},
        {"glucose": 120.0, "carbs": 45.0, "bolus": 3.0}
    ],
    "predictions": [
        {"model": "nn", "horizon": "30min", "glucose": 135.0, "confidence": 0.85},
        {"model": "ode", "horizon": "30min", "glucose": 132.0, "confidence": 0.90}
    ]
}

# Запускаем логику 
print(" Запуск медицинской логики...")
result = process_recommendation_request(test_input)

print(f" Результат: Шаблон {result['template_id']}, Срочность: {result['urgency']}")
print(f" Текст: {result['recommendation_text']}")

# Отправляем результат 
print(" Отправка на бэкенд...")
send_recommendation_to_backend(result)
```
